# Whisper small: speech to text in your Google Colab

Runs on Google's free GPU in **your** Google account, so nothing runs on your phone or laptop.

1. **Runtime > Change runtime type > T4 GPU** (faster; CPU works too).
2. Run the cells top to bottom (the play button on each).
3. In step 3, pick an audio file (voice note, mp3, wav, m4a). Run it again for another file.

```
#@title 1. Keep the model in your Google Drive (optional)
#@markdown Saves the download in `MyDrive/ML-HOST/models`, so next time it loads without downloading again. Untick to skip Drive.
use_drive = True #@param {type:"boolean"}
import os
cache = "/content/hf"
if use_drive:
    from google.colab import drive
    drive.mount("/content/drive")
    cache = "/content/drive/MyDrive/ML-HOST/models"
os.makedirs(cache, exist_ok=True)
os.environ["HF_HOME"] = cache  # must be set before transformers is imported
```

```
#@title 2. Load the model
import torch
from transformers import pipeline
cuda = torch.cuda.is_available()
asr = pipeline("automatic-speech-recognition", model="openai/whisper-small", device=0 if cuda else -1)
print("Loaded on", "GPU" if cuda else "CPU (slow: Runtime > Change runtime type > T4 GPU)")
```

```
#@title 3. Transcribe an audio file
from google.colab import files
for name in files.upload():
    print(name, "->", asr(name, chunk_length_s=30)["text"])
```
